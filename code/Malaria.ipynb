```python
import pandas as pd
import os
import re

PARASITIZED_CSV = "/content/patientid_cellmapping_parasitized.csv"
UNINFECTED_CSV = "/content/patientid_cellmapping_uninfected.csv"

PARASITIZED_DIR = "/content/malaria_project/data/cell_images/cell_images/Parasitized"
UNINFECTED_DIR = "/content/malaria_project/data/cell_images/cell_images/Uninfected"

parasitized_raw = pd.read_csv(PARASITIZED_CSV)
uninfected_raw = pd.read_csv(UNINFECTED_CSV)

print("Raw parasitized shape:", parasitized_raw.shape)
print("Raw uninfected shape:", uninfected_raw.shape)
```

```python
def convert_mapping_csv(df, label):
    records = []

    # First column contains the source/group ID
    group_column = df.columns[0]

    # Remaining columns contain image filenames
    for _, row in df.iterrows():

        group_id = str(row[group_column]).strip()

        for column in df.columns[1:]:

            value = row[column]

            if pd.isna(value):
                continue

            filename = str(value).strip()

            # Remove [, ], ', and "
            filename = filename.replace("[", "")
            filename = filename.replace("]", "")
            filename = filename.replace("'", "")
            filename = filename.replace('"', "")
            filename = filename.strip()

            if filename.lower().endswith(".png"):
                records.append({
                    "filename": filename,
                    "group_id": group_id,
                    "label": label
                })

    return pd.DataFrame(records)


parasitized_mapping = convert_mapping_csv(
    parasitized_raw,
    "Parasitized"
)

uninfected_mapping = convert_mapping_csv(
    uninfected_raw,
    "Uninfected"
)

mapping_df = pd.concat(
    [parasitized_mapping, uninfected_mapping],
    ignore_index=True
)

print("Total mapped images:", len(mapping_df))
display(mapping_df.head(10))
```

```python
import pandas as pd

train_df = pd.read_csv(
    "/content/malaria_project/data/splits/train.csv"
)

val_df = pd.read_csv(
    "/content/malaria_project/data/splits/validation.csv"
)

test_df = pd.read_csv(
    "/content/malaria_project/data/splits/test.csv"
)

print(train_df.shape)
print(val_df.shape)
print(test_df.shape)
```

```python
LABEL_MAP = {
    "Uninfected": 0,
    "Parasitized": 1
}

train_df["label_id"] = train_df["label"].map(LABEL_MAP)
val_df["label_id"] = val_df["label"].map(LABEL_MAP)
test_df["label_id"] = test_df["label"].map(LABEL_MAP)

print(train_df[["label", "label_id"]].drop_duplicates())
```

```python
import tensorflow as tf

IMG_SIZE = (128, 128)

def load_image(filepath, label):

    image = tf.io.read_file(filepath)

    image = tf.image.decode_png(
        image,
        channels=3
    )

    image = tf.image.convert_image_dtype(
        image,
        tf.float32
    )

    # Preserve aspect ratio and pad to 128x128
    image = tf.image.resize_with_pad(
        image,
        IMG_SIZE[0],
        IMG_SIZE[1]
    )

    return image, label
```

```python
BATCH_SIZE = 32

train_ds = tf.data.Dataset.from_tensor_slices(
    (
        train_df["filepath"].values,
        train_df["label_id"].values
    )
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (
        val_df["filepath"].values,
        val_df["label_id"].values
    )
)

test_ds = tf.data.Dataset.from_tensor_slices(
    (
        test_df["filepath"].values,
        test_df["label_id"].values
    )
)

train_ds = (
    train_ds
    .shuffle(
        buffer_size=len(train_df),
        seed=42,
        reshuffle_each_iteration=True
    )
    .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

val_ds = (
    val_ds
    .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

test_ds = (
    test_ds
    .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)
```

```python
import tensorflow as tf

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip(
        mode="horizontal_and_vertical"
    ),
    tf.keras.layers.RandomRotation(
        factor=0.08
    ),
    tf.keras.layers.RandomZoom(
        height_factor=0.10,
        width_factor=0.10
    ),
    tf.keras.layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    ),
    tf.keras.layers.RandomContrast(
        factor=0.10
    )
], name="data_augmentation")

print(data_augmentation)
```

```python
import matplotlib.pyplot as plt

images, labels = next(iter(train_ds))

sample_image = images[0:1]

plt.figure(figsize=(12, 8))

for i in range(12):

    augmented_image = data_augmentation(
        sample_image,
        training=True
    )[0]

    plt.subplot(3, 4, i + 1)

    plt.imshow(
        tf.clip_by_value(
            augmented_image,
            0.0,
            1.0
        )
    )

    label = (
        "Parasitized"
        if labels[0].numpy() == 1
        else "Uninfected"
    )

    plt.title(label)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from tensorflow import keras
from tensorflow.keras import layers

IMG_SIZE = (128, 128, 3)

baseline_model = keras.Sequential([

    layers.Input(shape=IMG_SIZE),

    # Data augmentation
    data_augmentation,

    # Feature extraction
    layers.Conv2D(
        32,
        (3, 3),
        padding="same",
        activation=None
    ),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.MaxPooling2D(),

    layers.Conv2D(
        64,
        (3, 3),
        padding="same",
        activation=None
    ),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.MaxPooling2D(),

    layers.Conv2D(
        128,
        (3, 3),
        padding="same",
        activation=None
    ),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.MaxPooling2D(),

    layers.Conv2D(
        256,
        (3, 3),
        padding="same",
        activation=None
    ),
    layers.BatchNormalization(),
    layers.ReLU(),

    # Classification
    layers.GlobalAveragePooling2D(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(0.4),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])

baseline_model.summary()
```

```python
baseline_model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss="binary_crossentropy",
    metrics=[
        keras.metrics.BinaryAccuracy(
            name="accuracy"
        ),
        keras.metrics.Precision(
            name="precision"
        ),
        keras.metrics.Recall(
            name="recall"
        ),
        keras.metrics.AUC(
            name="auc"
        )
    ]
)
```

```python
callbacks = [

    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),

    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-6,
        verbose=1
    ),

    keras.callbacks.ModelCheckpoint(
        "/content/malaria_project/models/baseline_cnn_best.keras",
        monitor="val_loss",
        save_best_only=True
    )
]
```

```python
history = baseline_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=callbacks
)
```

```python
# ============================================================
# BASELINE CNN — FINAL TEST EVALUATION
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras

# Evaluate on the untouched test set
test_results = baseline_model.evaluate(
    test_ds,
    verbose=1,
    return_dict=True
)

print("\n========== TEST RESULTS ==========")

for metric, value in test_results.items():
    print(f"{metric:12s}: {value:.4f}")
```

```python
# ============================================================
# GENERATE TEST PREDICTIONS
# ============================================================

y_true = []
y_prob = []

for images, labels in test_ds:
    predictions = baseline_model.predict(images, verbose=0)

    y_true.extend(labels.numpy())
    y_prob.extend(predictions.ravel())

y_true = np.array(y_true)
y_prob = np.array(y_prob)

# Default classification threshold
y_pred = (y_prob >= 0.5).astype(int)

print("Number of test samples:", len(y_true))
print("Number of predictions :", len(y_pred))
```

```python
# ============================================================
# DETAILED CLASSIFICATION METRICS
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
auc = roc_auc_score(y_true, y_prob)

# Confusion matrix
tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

# Specificity
specificity = tn / (tn + fp)

# False positive rate
fpr_value = fp / (fp + tn)

# False negative rate
fnr_value = fn / (fn + tp)

print("\n========== FINAL TEST METRICS ==========")

print(f"Accuracy     : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision    : {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall       : {recall:.4f} ({recall*100:.2f}%)")
print(f"Specificity  : {specificity:.4f} ({specificity*100:.2f}%)")
print(f"F1 Score     : {f1:.4f} ({f1*100:.2f}%)")
print(f"ROC-AUC      : {auc:.4f} ({auc*100:.2f}%)")

print("\n========== CONFUSION MATRIX VALUES ==========")

print(f"True Negatives  (TN): {tn}")
print(f"False Positives (FP): {fp}")
print(f"False Negatives (FN): {fn}")
print(f"True Positives  (TP): {tp}")

print("\n========== ERROR RATES ==========")

print(f"False Positive Rate: {fpr_value:.4f} ({fpr_value*100:.2f}%)")
print(f"False Negative Rate: {fnr_value:.4f} ({fnr_value*100:.2f}%)")
```
